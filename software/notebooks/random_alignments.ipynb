```python
import numpy as np
import probe
import trimesh as tm
```

```python
cube = tm.creation.box(extents=(0.1, 0.1, 0.1))
```

```python
cup = probe.Probe()

T_nominal = np.eye(4)
trans_bounds = np.array([[-0.06, 0.06], [-0.06, 0.06], [0.05, 0.1]])

T = probe.sample_pose(T_nominal, trans_bounds)
cup.transform(T)
```

```python
lines = cup.make_ray_lines()
cup_outline = cup.make_probe()

scene = tm.Scene()
scene.add_geometry(cube, node_name="small_cube")
for ray in lines:
    ray.colors = [[255, 0, 0, 255]]
    scene.add_geometry(ray)
cup_outline.colors = [[0, 255, 0, 100]]
scene.add_geometry(cup_outline, node_name="probe_outline")
scene.show()
```

```python
cup.sense(cube)
```
